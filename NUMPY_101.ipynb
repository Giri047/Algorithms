### List Vs Numpy

```python
import numpy as np

py_list= [1,2,4,5]
print("Multiplying list: ", py_list*2) # element wise multiplication
np_arr = np.array(py_list)
print("Multiplying numpy array: ", np_arr*2)
```

```python
# Checking which is more optimized during runtime
import time

start_time = time.time()
py_list = [i*2 for i in range(1000000)]
print("List operation time",time.time() - start_time)

start_time = time.time()
np_arr = np.arange(1000000) * 2
print("Numpy operation time",time.time() - start_time)



# start_time = time.time()
# np_arr = np_arr * 2
# end_time = time.time()
# print(f"Time taken for numpy array multiplication: {end_time - start_time} seconds")
```
